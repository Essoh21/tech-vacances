```python
import requests
from bs4 import BeautifulSoup

url = "https://github.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
tous_les_grands_titres = soup.find_all(["h1", "h2", "h3", "h4", "h5", "h6"])
les_titres = [titre.text for titre in tous_les_grands_titres]
for titre in les_titres:
    print(titre.strip())

```
